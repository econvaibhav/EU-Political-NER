```python
import openai
import pandas as pd
import json
from tqdm import tqdm
from requests.exceptions import Timeout

openai.api_key = os.environ["OPENAI_API_KEY"]
input_file = r"C:\project\political_ner\alex_corrections.json"

SYSTEM_PROMPT = (
    "You are an assistant trained to resolve name variations and suggest corrections. "
    "Your role is to ensure the most accurate and appropriate spelling of names in their original language, including symbols, while resolving duplicate options."
)

USER_PROMPT = "Are you clear about your role?"
ASSISTANT_PROMPT = (
    "Yes, I understand. My task is to: "
    "1. Choose the correct spelling for each name, prioritizing the original language with accurate diacritics or symbols. "
    "2. Resolve duplicates by removing unnecessary brackets or alternative spellings if they convey the same meaning. "
    "3. Return a JSON object containing the cleaned and corrected names."
)

def openai_chat_completion_response(sentence, timeout=10):
    try:
        user_prompt = (
            f"Given these name variations: {sentence}\n"
            "1) Choose the most appropriate name in its original language with correct diacritics or symbols.\n"
            "2) Remove duplicates or alternative spellings if they convey the same meaning.\n"
            "3) Return the final result as valid JSON with the key 'CorrectedName', e.g.:\n"
            "{\"CorrectedName\": [\"Your Corrected Name Here\"]}.\n"
        )

        response = openai.chat.completions.create(
            model="gpt-3.5-turbo",  # or "gpt-4", etc.
            messages=[
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": USER_PROMPT},
                {"role": "assistant", "content": ASSISTANT_PROMPT},
                {"role": "user", "content": user_prompt},
            ],
            timeout=timeout,
            temperature=0,
        )
        
        # Extract the assistant's message content
        raw_text = response.choices[0].message.content.strip()
        
        # Attempt to parse as JSON
        parsed_json = json.loads(raw_text)
        return parsed_json

    except Timeout:
        return {"CorrectedName": ["Error: Timeout"]}
    except Exception as e:
        # If something goes wrong, return the error in JSON
        return {"CorrectedName": [f"Error: {e}"]}

def main():
    # Load input data
    with open(input_file, "r", encoding="utf-8") as file:
        data = json.load(file)

    results = []

    # Process rows with progress bar
    for key, names in tqdm(data.items(), desc="Processing rows"):
        sentence = " & ".join(names)  # Combine names into a single prompt
        response_dict = openai_chat_completion_response(sentence)

        corrected_name = ", ".join(response_dict.get("CorrectedName", []))

        results.append({
            "Original": names,
            "CorrectedName": corrected_name
        })

    # Save results to a JSON file
    output_json_path = r"C:\project\political_ner\output_results.json"
    with open(output_json_path, "w", encoding="utf-8") as json_file:
        json.dump(results, json_file, ensure_ascii=False, indent=4)

    # Save results to a CSV file
    output_csv_path = r"C:\project\political_ner\output_results.csv"
    results_df = pd.DataFrame(results)
    results_df.to_csv(output_csv_path, index=False, encoding="utf-8")

    print(f"Processing complete. Results saved to '{output_json_path}' and '{output_csv_path}'.")

if __name__ == "__main__":
    main()
```
